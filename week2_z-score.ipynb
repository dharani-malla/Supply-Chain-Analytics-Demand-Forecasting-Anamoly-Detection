# EXPLORATORY STATISTICAL ANALYSIS

```python
numerical_columns = [
    "Inventory Level",
    "Units Sold",
    "Units Ordered",
    "Demand Forecast",
    "Price",
    "Discount",
    "Competitor Pricing"
]

df[numerical_columns].head()
```

# Z-SCORE BASED ANOMALY DETECTION

```python
z_scores = df[numerical_columns].apply(zscore)
```

```python
z_scores.head()
```

```python
threshold = 3

df["Anomaly"] = (np.abs(z_scores) > threshold).any(axis=1)
```
